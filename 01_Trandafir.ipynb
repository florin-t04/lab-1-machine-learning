### Exercițiul 1: Încărcarea și analiza inițială a datelor

```python
import numpy as np
import pandas as pd

data = pd.read_csv("Iris_Data.csv")
data.head()
```

```python
# nr coloane
print(data.shape[0])
# nume coloane
print(data.columns.tolist())
# Data types
print(data.dtypes)
```

### Exercițiul 2: Curățarea și transformarea datelor text


```python
print(data.species)
data.head()
```

```python
data['species'] = data.species.str.replace('Iris-', '')
data.head()
```

### Exercițiul 3: Statistica descriptivă și amplitudinea (range)


```python
stats = data.describe()
stats.loc["range"] = stats.loc["max"] - stats.loc["min"]
stats
```

### Exercițiul 4: Grupatarea datelor pe specii (Media și Mediana)


```python
data.groupby('species').mean()
data.groupby('species').median()
```

### Exercițiul 5: Agregare avansată utilizând un dicționar


```python
data.groupby('species').agg({ 
    "sepal_length": ["mean", "median"],
    "sepal_width": ["mean", "median"],
    "petal_width": ["mean", "median"],
    "petal_length": "max",
})
```

### Exercițiul 6: Vizualizarea datelor cu ajutorul Matplotlib


```python
import matplotlib.pyplot as plt
```

```python
#x=sepal_length y= sepal_width
ax=plt.axes()
ax.scatter(data.sepal_length, data.sepal_width)
ax.set(xlabel='sepal_length', ylabel='sepal_width', title='Scatter Plot of Sepal length vs Sepal width')
```
